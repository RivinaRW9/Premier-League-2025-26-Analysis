# Create Teams Table
Create a table to store Premier League clubs and their unique Team IDs.

```python
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS Teams (
    TeamID INTEGER PRIMARY KEY,
    TeamName TEXT NOT NULL
)
""")

conn.commit()

print("Teams table created!")
```

# Create Players Table 
Create a table to store player information including name, team, position, age and nationality.

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Players (
    PlayerID INTEGER PRIMARY KEY,
    PlayerName TEXT NOT NULL,
    TeamID INTEGER NOT NULL,
    Position TEXT,

    FOREIGN KEY (TeamID)
    REFERENCES Teams(TeamID)
)
""")

conn.commit()

print("Players table created!")
```

# SQL analysis
Run sql quaries to analyse player performance and generate football insights.

# Query 1 - Top Goalscorers


```python
top_goals_query = """
SELECT
    p.PlayerName,
    ps.Goals
FROM Players AS p
JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID
ORDER BY ps.Goals DESC
LIMIT 10;
"""
```

```python
top_goals_df = pd.read_sql_query(top_goals_query, conn)

top_goals_df
```

# Query 2 - Top Assist Providers

```python
top_assists_query = """
SELECT
    p.PlayerName,
    ps.Assists
FROM Players AS p
JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID
ORDER BY ps.Assists DESC
LIMIT 10;
"""
```

```python
top_assists_df = pd.read_sql_query(top_assists_query, conn)

top_assists_df
```

# Query 3 - Team Discipline Score

```python
team_discipline_query = """
SELECT
    t.TeamName,
    SUM(ps.YellowCards) AS YellowCards,
    SUM(ps.RedCards) AS RedCards,
    SUM(ps.YellowCards + (ps.RedCards * 2)) AS DisciplineScore
FROM Teams AS t
JOIN Players AS p
    ON t.TeamID = p.TeamID
JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID
GROUP BY t.TeamName
ORDER BY DisciplineScore DESC;
"""
```

```python
team_discipline_df = pd.read_sql_query(
    team_discipline_query,
    conn
)

team_discipline_df
```

# Query 4 - Average age of each team

```python
average_age_by_team_query = """
SELECT
    t.TeamName,
    ROUND(AVG(p.Age), 1) AS AverageAge
FROM Teams AS t
JOIN Players AS p
    ON t.TeamID = p.TeamID
GROUP BY t.TeamName
ORDER BY AverageAge DESC;
"""
```

```python
average_age_by_team_df = pd.read_sql_query(
    average_age_by_team_query,
    conn
)

average_age_by_team_df
```

# Query 5 - Total goals scored by each team


```python
goals_by_team_query = """
SELECT
    t.TeamName,
    SUM(ps.Goals) AS TotalGoals
FROM Teams AS t
JOIN Players AS p
    ON t.TeamID = p.TeamID
JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID
GROUP BY t.TeamName
ORDER BY TotalGoals DESC;
"""
```

```python
goals_by_team_df = pd.read_sql_query(
    goals_by_team_query,
    conn
)

goals_by_team_df
```

# Query 6 - Players with the most goal contributions per 90 (G/A)

```python
goal_contributions_per_90_query = """
SELECT
    p.PlayerName,
    ps.Goals,
    ps.Assists,
    ps.Nineties,
    ROUND((ps.Goals + ps.Assists) / ps.Nineties, 2) AS GoalContributionsPer90
FROM Players AS p
JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID
WHERE ps.Nineties >= 10
ORDER BY GoalContributionsPer90 DESC
LIMIT 10;
"""
```

```python
goal_contributions_per_90_df = pd.read_sql_query(
    goal_contributions_per_90_query,
    conn
)

goal_contributions_per_90_df
```

# Query 7 - Players with the highest percentage of team goals.

```python
team_goals_query = """
SELECT
    p.TeamID,
    SUM(ps.Goals) AS TeamGoals
FROM Players AS p
JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID
GROUP BY p.TeamID;
"""
```

```python
team_goals_df = pd.read_sql_query(
    team_goals_query,
    conn
)

team_goals_df
```

```python
percentage_team_goals_query = """
SELECT
    p.PlayerName,
    t.TeamName,
    ps.Goals,
    tg.TeamGoals,
    ROUND((ps.Goals * 100.0) / tg.TeamGoals, 1) AS GoalPercentage
FROM Players AS p

JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID

JOIN Teams AS t
    ON p.TeamID = t.TeamID

JOIN (
    SELECT
        p.TeamID,
        SUM(ps.Goals) AS TeamGoals
    FROM Players AS p
    JOIN PlayerStats AS ps
        ON p.PlayerID = ps.PlayerID
    GROUP BY p.TeamID
) AS tg
    ON p.TeamID = tg.TeamID

WHERE ps.Nineties >= 10

ORDER BY GoalPercentage DESC
LIMIT 10;
"""
```

```python
percentage_team_goals_df = pd.read_sql_query(
    percentage_team_goals_query,
    conn
)

percentage_team_goals_df
```

# Query 8 - Total number of assists by each team

```python
assists_by_team_query = """
SELECT
    t.TeamName,
    SUM(ps.Assists) AS TotalAssists
FROM Teams AS t
JOIN Players AS p
    ON t.TeamID = p.TeamID
JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID
GROUP BY t.TeamName
ORDER BY TotalAssists DESC;
"""
```

```python
assists_by_team_df = pd.read_sql_query(
    assists_by_team_query,
    conn
)

assists_by_team_df
```

# Query 9 - Top Players with the most goal contributions

```python
goal_contributions_query = """
SELECT
    p.PlayerName,
    ps.Goals,
    ps.Assists,
    (ps.Goals + ps.Assists) AS GoalContributions
FROM Players AS p
JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID
ORDER BY GoalContributions DESC
LIMIT 10;
"""
```

```python
goal_contributions_df = pd.read_sql_query(
    goal_contributions_query,
    conn
)

goal_contributions_df
```

# Query 10 - Players with the highest Goals per 90

```python
goals_per_90_query = """
SELECT
    p.PlayerName,
    ps.Goals,
    ps.Nineties,
    ROUND(ps.Goals / ps.Nineties, 2) AS GoalsPer90
FROM Players AS p
JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID
WHERE ps.Nineties >= 10
ORDER BY GoalsPer90 DESC
LIMIT 10;
"""
```

```python
goals_per_90_df = pd.read_sql_query(
    goals_per_90_query,
    conn
)

goals_per_90_df
```

# Query 11 - Players with the most minutes played

```python
minutes_played_query = """
SELECT
    p.PlayerName,
    t.TeamName,
    ps.MinutesPlayed
FROM Players AS p
JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID
JOIN Teams AS t
    ON p.TeamID = t.TeamID
ORDER BY ps.MinutesPlayed DESC
LIMIT 10;
"""
```

```python
minutes_played_df = pd.read_sql_query(
    minutes_played_query,
    conn
)

minutes_played_df
```

# Query 12 - Top Goalscorers from each team

```python
top_scorer_each_team_query = """
SELECT
    t.TeamName,
    p.PlayerName,
    ps.Goals
FROM Players AS p
JOIN PlayerStats AS ps
    ON p.PlayerID = ps.PlayerID
JOIN Teams AS t
    ON p.TeamID = t.TeamID
WHERE ps.Goals = (
    SELECT MAX(ps2.Goals)
    FROM Players AS p2
    JOIN PlayerStats AS ps2
        ON p2.PlayerID = ps2.PlayerID
    WHERE p2.TeamID = p.TeamID
)
ORDER BY t.TeamName;
"""
```

```python
top_scorer_each_team_df = pd.read_sql_query(
    top_scorer_each_team_query,
    conn
)

top_scorer_each_team_df
```
